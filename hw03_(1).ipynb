```python
#preprocessing
import spacy
#!spacy download ru_core_news_md
nlp=spacy.load("ru_core_news_md")
texts_name = [f'{x}{i}.txt' for x in ['мат', 'астроном', 'астро', 'палео'] for i in range(1, 5)]
texts = {}
for text in texts_name:
    with open(text, 'r', encoding='utf-8') as file:
      text_i = file.read()
      text_t = [token.lemma_ for token in nlp(text_i) if (token.is_stop != True and token.is_alpha == True)]
      name = text.rstrip('.txt').rstrip('1234')
      if name not in texts:
        texts[name] = set()
      texts[name].update(text_t)
```

```python
#составление словаря уникальных слов для каждой тематики
text_type_dict = {}
for text_type_base in texts:
  difference = texts[text_type_base]
  for text_type_comp in texts:
    if texts[text_type_comp] != texts[text_type_base]:
      difference = difference - texts[text_type_comp]
  text_type_dict[text_type_base] = difference
```

```python
import nltk
from nltk import word_tokenize
from nltk.text import Text
import spacy

def analyze_preposition(text_path, preposition):

    with open(text_path, 'r', encoding='utf-8') as file:
        text_data = file.read()
    tokens = word_tokenize(text_data)
    text_T = Text(tokens)

    con_list = text_T.concordance_list(preposition, lines=100000)

    nlp = spacy.load('ru_core_news_md')

    preposition_dictionary = {}
    docs = list(nlp.pipe([element.line for element in con_list]))
    for i, doc in enumerate(docs):
        sentence = con_list[i].line
        found_prep = False

        for token in doc:
          if found_prep:
            if token.text in ['!', '?', '.', ',', ':', ';', '...']:
              break
            if token.pos_ in ['NOUN', 'ADJ', 'PRON']:
              casei = token.morph.get('Case')
              if casei:
                case_key = casei[0]
                if case_key not in preposition_dictionary:
                  preposition_dictionary[case_key] = [0, []]
                preposition_dictionary[case_key][0] += 1
                if len(preposition_dictionary[case_key][1]) < 5:
                  preposition_dictionary[case_key][1].append(sentence)
              break
          elif token.text.lower() == preposition:
            found_prep = True

    return preposition_dictionary
```

```python
#!pip install pyconll
import pyconll
text = pyconll.load_from_file('GramEval2020-17cent-train.conllu')

for sentence in text:
  flag_i = False
  for token in sentence:
    if token.deprel in ['discourse', 'dislocated']:
      flag_i = True
      sentence_f = ' '.join(token.form for token in sentence)
      print(f"Предложение:\n{sentence_f}")
      print(f"Зависимость: {token.deprel}, Слово: '{token.form}'")
      print('__________________________________')
  if flag_i == False:
    print('Теги discourse, dislocated не были найдены в файле')
```

```python
import pyconll
def search_tag(text_path, *tags):
  text = pyconll.load_from_file(text_path)
  flag_i = False
  for sentence in text:
    for token in sentence:
      if token.deprel in tags:
        flag_i = True
        sentence_f = ' '.join(token.form for token in sentence)
        print(f"Предложение:\n{sentence_f}")
        print(f"Зависимость: {token.deprel}, Слово: '{token.form}'")
        print('__________________________________')
  if flag_i == False:
    print(f'Теги {tags} не были найдены в файле')
  return
```
